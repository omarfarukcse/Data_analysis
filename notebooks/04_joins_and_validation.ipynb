## Example: customers + orders

## Validate keys before merge

```python
assert_no_duplicate_keys(customers, ['customer_id'])
# orders may have multiple rows per customer_id (that is OK)
merged = orders.merge(customers, on='customer_id', how='left', validate='many_to_one')
merged.head()
```
