# 03 — KPI Tables (groupby + pivot_table)

Goal: produce dashboard-ready KPI tables (one row per time period + dimensions).

```python
import pandas as pd
from src.kpi import add_month_column, kpi_groupby, pivot_kpis
```

```python
df = pd.read_csv("../data/sample.csv")  # change me
df.head()
```

## Example KPI build
Assumes columns like `order_date`, `region`, `order_id`, `revenue` (change as needed).

```python
df2 = add_month_column(df, 'order_date')
kpis = kpi_groupby(
    df2,
    dims=['month', 'region'],
    metrics={
        'revenue': ('revenue', 'sum'),
        'orders': ('order_id', 'nunique'),
    },
)
kpis.head()
```

```python
kpis_wide = pivot_kpis(
    kpis,
    index=['month'],
    columns='region',
    values=['revenue', 'orders'],
)
kpis_wide.head()
```
